# Figures and results

Draws each figure from the model runs in `results/` and writes the tables and summary values. Figures are saved to
`outputs/` as PNG (300 dpi) and PDF. Run `2_run_scenarios.ipynb` first.

## Setup

```python
import os

import geopandas as gpd
from IPython.display import Image, display

import analysis as m

RESULTS, OUT = 'results', 'outputs'
os.makedirs(OUT, exist_ok=True)


def show(name, width=900):
    display(Image(filename=os.path.join(OUT, name + '.png'), width=width))
```

## Data

```python
d = m.load(RESULTS)                                   # results per settlement, without and with PUE
bounds = gpd.read_file('data/meru_subcounties.geojson')
v = m.compute_values(d)                               # summary values
via, central = m.minigrid_viability(RESULTS)          # mini-grid viability by household threshold, and by settlement
oat = m.capital_cost_sensitivity(RESULTS)             # ±20% capital-cost sensitivity (None if the runs are missing)
```

## Tables and summary values

```python
m.write_tables(RESULTS, OUT)
```

## Base-year (2020) electrification and solar resource

```python
m.fig_baseline_electrification(d['PUE'], bounds, OUT)
show('baseline_electrification', 750)
```

## Demand locations for health, water, crop farming and poultry

```python
m.fig_pue_demand_maps(d['PUE'], bounds, OUT)
show('pue_demand_maps')
```

## New electricity demand and required capacity

```python
m.fig_demand_and_capacity(v, OUT, d)
show('demand_and_capacity', 600)
```

## Least-cost technology in 2030, without and with PUE

```python
m.fig_technology_maps(d, bounds, OUT)
show('technology_maps')
```

## Rural investment per kW, grid and off-grid

```python
m.fig_cost_per_kw(v, OUT)
show('cost_per_kw_rural', 600)
```

## Mini-grid viability in off-grid settlements with PUE

```python
m.fig_minigrid_viability(central, bounds, OUT, via)
show('minigrid_viability', 700)
```

## Capital-cost sensitivity (±20%)

```python
if oat is not None:
    m.fig_capital_cost_sensitivity(oat, OUT)
    show('capital_cost_sensitivity')
    m.fig_capital_cost_sensitivity_all(oat, OUT)
    show('capital_cost_sensitivity_all', 1000)
```

## Mini-grid sizes by anchor load

```python
m.fig_minigrid_sizes(d['PUE'], OUT)
show('minigrid_sizes', 600)
```
