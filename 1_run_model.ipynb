# OnSSET model run for Meru County, Kenya

Least-cost electrification of Meru County to 2030, with and without productive-use (PUE) and community-service demand.
The notebook is based on the GEP Generator interface to [OnSSET](http://www.onsset.org/) (KTH dESA).

Choose the scenario in the scenario cell below the imports and use Run All. Results are written to `results/<scenario>/`.
To run every scenario, each in a fresh kernel, use `2_run_scenarios.ipynb`.

Main settings:
- Scenarios: without PUE (`data/onsset_no_pue.csv`), with PUE (`data/onsset_with_pue.csv`), and sensitivity runs on the
  mini-grid cost, PUE uptake and each capital cost (±20%).
- Settlements with missing raster data take the values of their nearest neighbour.
- Health, water and poultry demand is added to the settlement load; crop farming is costed separately as farm-level solar.
- Rural Tier 2, urban Tier 4. Stand-alone PV up to 200 W per household (Tier 2).
- PV mini-grid USD 4,000/kW including storage, USD 140 per connection; stand-alone PV USD 9,000/4,800/1,900/1,500 per kW
  by size band.
- PUE demand added once, in 2030.
- Diesel, wind and hybrid mini-grids excluded.
- Grid-extension cost recorded for every settlement, for the mini-grid comparison.

```python
import os

# Scenario selection. The scenarios differ only in the settings listed here.
# 2_run_scenarios.ipynb runs each one in a fresh kernel by setting ONSSET_SCENARIO;
# when running this notebook by hand, change the default name below.
SCENARIOS = {
    # Scenario 1 (baseline): household demand only
    'Scenario1_baseline_noPUE': {'input_file': 'data/onsset_no_pue.csv',   'sa_pv_capital_cost_5': 999999999},
    # Scenario 2 (enhanced demand): household + PUE and community-service demand
    'Scenario2_enhanced_PUE':   {'input_file': 'data/onsset_with_pue.csv', 'sa_pv_capital_cost_5': 999999999},
    # Sensitivity runs (enhanced-demand scenario):
    # mini-grid cost x0.625 (USD 2,500/kW) and x1.3 (USD 5,200/kW), the range of Kenyan productive-use mini-grid costs
    # (AMDA 2025); PUE demand realised at 50% of the planned values.
    'Sens_MGcostLow_PUE':  {'input_file': 'data/onsset_with_pue.csv', 'sa_pv_capital_cost_5': 999999999, 'mg_cost_factor': 0.625},
    'Sens_MGcostHigh_PUE': {'input_file': 'data/onsset_with_pue.csv', 'sa_pv_capital_cost_5': 999999999, 'mg_cost_factor': 1.3},
    'Sens_PUE50':          {'input_file': 'data/onsset_with_pue.csv', 'sa_pv_capital_cost_5': 999999999, 'pue_scale': 0.5},  # central mini-grid cost
}
# sa_pv_capital_cost_5 = 999999999 makes stand-alone PV above 200 W per household (above Multi-Tier Framework Tier 2)
# unaffordable in all scenarios, so settlements needing more than Tier 2-sized systems are served by the grid or a mini-grid.

# Crop-farming solutions (EducationDemand column) are farm-level solar irrigation systems (0.62 or 1 kWp per farm).
# They are taken out of the cluster load and costed separately in analysis.py (stand-alone solar at the
# 101-200 W price, USD 1,500/kWp), so they do not drive the settlement's technology choice.
crop_farming_standalone = True

# GridCellArea in the input files is in square degrees x 1e-6. The values are used as given (False);
# True converts them to km2.
fix_grid_cell_area = False

# One-at-a-time capital-cost sensitivity (enhanced-demand scenario): each run scales one capital cost by 0.8 or 1.2.
# The names are used by cf() in the cost cells below.
OAT_PARAMETERS = ['sa_pv_1', 'sa_pv_2', 'sa_pv_3', 'sa_pv_4', 'mg_pv', 'mg_connection', 'mg_diesel', 'mg_wind', 'sa_diesel',
                  'grid_connection', 'mv_line', 'lv_line', 'hv_line', 'transformers', 'grid_capacity']
for _p in OAT_PARAMETERS:
    for _f in (0.8, 1.2):
        SCENARIOS['OAT_%s_%d' % (_p, round(_f * 100))] = {**SCENARIOS['Scenario2_enhanced_PUE'], 'cost_factors': {_p: _f}}

scenario_name = os.environ.get('ONSSET_SCENARIO', 'Scenario2_enhanced_PUE')
scenario = SCENARIOS[scenario_name]
print('Running scenario:', scenario_name, scenario)

cost_factors = scenario.get('cost_factors', {})
def cf(name):
    """Capital-cost multiplier for the one-at-a-time sensitivity (1 = central value)."""
    return cost_factors.get(name, 1.0)
```

# 1. GIS data selection

The input file is set by the scenario (`data/onsset_no_pue.csv` or `data/onsset_with_pue.csv`).


```python
input_file = os.path.abspath(scenario['input_file'])

onsseter = SettlementProcessor(input_file)

# Keep the GIS attributes only (the input files also hold columns computed by an earlier model run) and rename lon/lat to X_deg/Y_deg.
_raw_cols = onsseter.df.columns[:onsseter.df.columns.get_loc('PopStartYear')].tolist()
onsseter.df = onsseter.df[_raw_cols]
onsseter.df.rename(columns={'lon': 'X_deg', 'lat': 'Y_deg'}, inplace=True)

# Reset PerCapitaDemand so residential demand is computed from the tier settings below.
onsseter.df['PerCapitaDemand'] = 0

# 160 settlements have no raster data (GHI, Elevation, Slope, TravelHours and WindVel all 0).
# Fill each missing value from the nearest settlement that has data.
import scipy.spatial
_raster_cols = ['GHI', 'Elevation', 'Slope', 'LandCover', 'TravelHours', 'WindVel']
_missing = onsseter.df['GHI'] == 0
if _missing.any():
    _ok = onsseter.df.index[~_missing]
    _tree = scipy.spatial.cKDTree(onsseter.df.loc[_ok, ['X_deg', 'Y_deg']].values)
    _, _nearest = _tree.query(onsseter.df.loc[_missing, ['X_deg', 'Y_deg']].values)
    onsseter.df.loc[_missing, _raster_cols] = onsseter.df.loc[_ok[_nearest], _raster_cols].values
print('Filled raster data for {} settlements from their nearest neighbour'.format(int(_missing.sum())))

if fix_grid_cell_area:
    onsseter.df['GridCellArea'] = onsseter.df['GridCellArea'] * 1e6 * 111.32 * 111.32 * np.cos(np.radians(onsseter.df['Y_deg']))
    print('GridCellArea converted to km2: total settled area {:.0f} km2'.format(onsseter.df['GridCellArea'].sum()))

# PUE uptake sensitivity: scale all planned PUE and community-service demand (1 = as planned)
pue_scale = scenario.get('pue_scale', 1.0)
for _c in ['HealthDemand', 'EducationDemand', 'CommercialDemand', 'AgriDemand']:
    onsseter.df[_c] = onsseter.df[_c] * pue_scale

if crop_farming_standalone:
    print('Crop farming taken out of the cluster load: {:.2f} GWh/yr'.format(onsseter.df['EducationDemand'].sum() / 1e6))
    onsseter.df['EducationDemand'] = 0
```

#### Lever 2: Electricity demand target
For the second lever, enter the target tier (level of electricity access) for urban and rural households respectively. *On the GEP Explorer, the following three electricity demand targets are used:*

*Top down demand target - Low: In this case, all urban clusters are tergeted to reach the current average consumption level of electrified households, and rural settlements are assigned Tier 1.* 

*Top down demand target - High: In this case, all urban clusters are tergeted to reach one Tier higher than the current average consumption level of electrified households, and rural settlements are assigned Tier 3.*

*Bottom up demand target (Poverty - GDP): In this case each settlement is assigned a demand target based on poverty and GDP levels as described in the methodology above. Choose "6" for both the urban_target_tier and rural_target_tier to use this option*

#### Lever 6: Rollout plan
This lever reflects the electrification approach to be examined. On the GEP Explorer, there are currently two options in use:

**Nationwide Least Cost approach:** This options aims to achieve the electrification rate targets set for the intermediate and end year. For the years where the target is set below 100%, the algorithm prioritizes grid densification first (ramp up in already electrified clusters) then selection based on lowest invetsment cost per capita to choose which clusters to be electrified. Note! a generation capacity limit for the centralized grid is applied in order to keep grid expansion to sensible levels until the intermediate year. On the GEP Explorer, this follows historical trends for the countries in the region that managed the largest relative increase in recent years.*

**Forced grid approach:** This can take the following forms.

- Forced grid under a defined buffer zone (auto_intensification set equal to Xkm) & least cost approach outside of the buffer zone.

- Forced grid under a defined buffer zone (auto_intensification set equal to Xkm) & no electrification outside the buffer zone.

# 7. Start a scenario run, which calculate and compare technology costs for every settlement in the country

Based on the previous calculation this piece of code identifies the LCoE that every off-grid technology can provide, for each single populated settlement of the selected country. The cell then takes all the currently grid-connected points in the country, and looks at the points within a certain distance from them, to see if it is more economical to connect them to the grid, or to use one of the off-grid technologies calculated above. Once more points are connected to the grid, the process is repeated, so that new points close to those points might also be connected. This is repeated until there are no new points to connect to the grid.
