# Run all scenarios

Runs `1_run_model.ipynb` once per scenario, each in a fresh kernel, and saves the results and the executed notebook to
`results/<scenario>/`. Each run takes about 2–4 minutes.

```python
import os
import subprocess
import sys

MODEL = '1_run_model.ipynb'
MAIN = ['Scenario1_baseline_noPUE', 'Scenario2_enhanced_PUE', 'Sens_MGcostLow_PUE', 'Sens_MGcostHigh_PUE', 'Sens_PUE50']

# One-at-a-time capital-cost sensitivity: each capital cost at 80% and 120% of its central value
OAT_PARAMETERS = ['sa_pv_1', 'sa_pv_2', 'sa_pv_3', 'sa_pv_4', 'mg_pv', 'mg_connection', 'mg_diesel', 'mg_wind', 'sa_diesel',
                  'grid_connection', 'mv_line', 'lv_line', 'hv_line', 'transformers', 'grid_capacity']
OAT = ['OAT_%s_%d' % (p, f) for p in OAT_PARAMETERS for f in (80, 120)]


def run(names):
    for name in names:
        out_dir = os.path.join('results', name)
        os.makedirs(out_dir, exist_ok=True)
        print('Running', name, flush=True)
        subprocess.run([sys.executable, '-m', 'jupyter', 'nbconvert', '--to', 'notebook', '--execute', MODEL,
                        '--output-dir', out_dir, '--output', name + '_executed', '--ExecutePreprocessor.timeout=-1'],
                       env={**os.environ, 'ONSSET_SCENARIO': name}, check=True)
```

## Main scenarios and sensitivity runs

```python
run(MAIN)
```

## Capital-cost sensitivity (30 runs, about 1 hour)

```python
run(OAT)
```
